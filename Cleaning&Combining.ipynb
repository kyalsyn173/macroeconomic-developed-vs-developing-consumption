```python
consumption = pd.read_csv("real_final_Consumption.csv")
inflation = pd.read_csv("real_final__Inflation.csv")
gdp = pd.read_csv("real_final_GDP_per_capita.csv")
unemployment = pd.read_csv("real_final_Unemployment_rate.csv")
exchange = pd.read_csv("real_final_Exchange_Rate.csv")
saving = pd.read_csv("real_final_GrossSaving.csv")
```

```python
real_final_df = consumption.merge(inflation,on=["Country","Year"])
```

```python
real_final_df.tail()
```

```python
real_final_df=real_final_df.merge(gdp,on=["Country","Year"])
```

```python
real_final_df=real_final_df.merge(unemployment,on=["Country","Year"])
```

```python
real_final_df=real_final_df.merge(exchange,on=["Country","Year"])
```

```python
real_final_df=real_final_df.merge(saving,on=["Country","Year"])
```

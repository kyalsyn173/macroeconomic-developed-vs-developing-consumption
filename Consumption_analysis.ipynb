### Prediction

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

train2 = df[df["Year"] < 2015]
test2 = df[(df["Year"] >= 2015) & (df["Year"] <= 2019)]

features1 = ["Inflation","GDP_per_capita","Unemployment_rate","ExchangeRate","GrossSaving"]
target1 = "Consumption"

dev_train2 = train2[train2["Group"] == "Developed"]
dev_test2 = test2[test2["Group"] == "Developed"]

mod_dev2 = LinearRegression()
mod_dev2.fit(dev_train2[features1], dev_train2[target1])

pred_dev2 = mod_dev2.predict(dev_test2[features1])

dev_test_results2 = dev_test2[["Country","Year","Consumption"]].copy()
dev_test_results2["Predicted_Consumption"] = pred_dev2

mae_dev2 = mean_absolute_error(dev_test2[target1], pred_dev2)
rmse_dev2 = np.sqrt(mean_squared_error(dev_test2[target1], pred_dev2))
r2_dev2 = r2_score(dev_test2[target1], pred_dev2)

print("\n=== Developed (2015–2019) ===")
print("MAE:", mae_dev2)
print("RMSE:", rmse_dev2)
print("R2:", r2_dev2)
```

```python
devp_train2 = train2[train2["Group"] == "Developing"]
devp_test2 = test2[test2["Group"] == "Developing"]

mod_devp2 = LinearRegression()
mod_devp2.fit(devp_train2[features1], devp_train2[target1])

pred_devp2 = mod_devp2.predict(devp_test2[features1])

devp_test_results2 = devp_test2[["Country","Year","Consumption"]].copy()
devp_test_results2["Predicted_Consumption"] = pred_devp2

# Metrics
mae_devp2 = mean_absolute_error(devp_test2[target1], pred_devp2)
rmse_devp2 = np.sqrt(mean_squared_error(devp_test2[target1], pred_devp2))
r2_devp2 = r2_score(devp_test2[target1], pred_devp2)

print("\n=== Developing (2015–2019) ===")
print("MAE:", mae_devp2)
print("RMSE:", rmse_devp2)
print("R2:", r2_devp2)
```

### Lagged Variable

#### Developing

```python

import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

df = df.sort_values(["Country", "Year"])

df["Inflation_lag1"] = df.groupby("Country")["Inflation"].shift(1)
df["GDP_lag1"] = df.groupby("Country")["GDP_per_capita"].shift(1)
df["Unemployment_lag1"] = df.groupby("Country")["Unemployment_rate"].shift(1)
df["Exchange_lag1"] = df.groupby("Country")["ExchangeRate"].shift(1)
df["GrossSaving_lag1"] = df.groupby("Country")["GrossSaving"].shift(1)

lagged_features = [
    "Inflation_lag1",
    "GDP_lag1",
    "Unemployment_lag1",
    "Exchange_lag1",
    "GrossSaving_lag1"
]

lagged_target = "Consumption"

df_lagged = df.dropna()

lag1_train = df_lagged[df_lagged["Year"] < 2015]
lag1_test  = df_lagged[(df_lagged["Year"] >= 2015) & (df_lagged["Year"] <= 2019)]

lag1_devp_train = lag1_train[lag1_train["Group"] == "Developing"]
lag1_devp_test  = lag1_test[lag1_test["Group"] == "Developing"]

model_devp = LinearRegression()

model_devp.fit(
    lag1_devp_train[lagged_features],
    lag1_devp_train[lagged_target]
)

pred_devp = model_devp.predict(
    lag1_devp_test[lagged_features]
)

lag1_devp_results = lag1_devp_test[["Country", "Year", "Consumption"]].copy()
lag1_devp_results["Predicted_Consumption"] = pred_devp

print(lag1_devp_results.head())

mae_devp = mean_absolute_error(
    lag1_devp_test[lagged_target],
    pred_devp
)

rmse_devp = np.sqrt(mean_squared_error(
    lag1_devp_test[lagged_target],
    pred_devp
))

r2_devp = r2_score(
    lag1_devp_test[lagged_target],
    pred_devp
)

print("\n=== Developing (Lag1, 2015–2019) ===")
print("MAE:", mae_devp)
print("RMSE:", rmse_devp)
print("R2:", r2_devp)
```

```python
lag1_2015_train = df_lagged[df_lagged["Year"] < 2015]
lag1_2015_test = df_lagged[(df_lagged["Year"] >= 2015) & (df_lagged["Year"] <= 2019)]

lag1_2015_dev_train = lag1_2015_train[lag1_2015_train['Group'] == 'Developed']
lag1_2015_dev_test = lag1_2015_test[lag1_2015_test['Group'] == 'Developed']

lag1_2015_mod_dev = LinearRegression()
lag1_2015_mod_dev.fit(lag1_2015_dev_train[lagged_features],lag1_2015_dev_train[lagged_target])

lag1_2015_pred_dev = lag1_2015_mod_dev.predict(lag1_2015_dev_test[lagged_features])

lag1_2015_dev_results = lag1_2015_dev_test[['Country','Year','Consumption']].copy()
lag1_2015_dev_results['Predicted_Consumption'] = lag1_2015_pred_dev


lag1_2015_mae_dev = mean_absolute_error(lag1_2015_dev_test[lagged_target],lag1_2015_pred_dev)

lag1_2015_rmse_dev = np.sqrt( mean_squared_error(lag1_2015_dev_test[lagged_target], lag1_2015_pred_dev))

lag1_2015_r2_dev = r2_score(lag1_2015_dev_test[lagged_target],lag1_2015_pred_dev)

print("\n=== Developed (Lag1, 2015–2019) ===")
print("MAE:", lag1_2015_mae_dev)
print("RMSE:", lag1_2015_rmse_dev)
print("R2:", lag1_2015_r2_dev)
```
